```python
df['floors'].mean
# df['floors'].median()
# df['floors'].mod()[0]
```

```python
# ' ' , N/A , None, Nan
df['floors'].fillna(df['floors'].mean(), inplace=True)
```

```python
def fillNaFloatMean(columns):
    for i in columns:
        if df[i].dtype in ['float64', 'int64']:  # only numeric
            df[i].fillna(df[i].mean(), inplace=True)
            print(f"Filled NaN in {i} with mean: {df[i].mean()}")
        else:
            print(f"Skipped {i} (non-numeric type: {df[i].dtype})")

columns = ['price', 'sqft_lot', 'sqft_living', 'condition', 'sqft_above', 'sqft_basement']
fillNaFloatMean(columns)

```

```python
# df['price'] = df['price'].astype('int64')

def ChangeFloatToInt(cols):
    for i in cols:
        # print(type(df[i]))
        df[i] = df[i].astype('int64')

columns = ['sqft_lot', 'sqft_living', 'condition', 'sqft_above', 'sqft_basement', 'bedrooms',
           'bathrooms']

ChangeFloatToInt(columns)
```

```python
df.info()
```

```python
# Convert specific object columns to integer (if numeric strings)

# Enter your object column names here
object_columns = ['date' , 'street' , 'city' ,'statezip' , 'country']  # e.g., ['sqft_living', 'price', 'sqft_lot']

for col in object_columns:
    df[col] = pd.to_numeric(df[col], errors='coerce').astype('Int64')

print("Conversion completed. Columns converted to int:", object_columns)
```

```python
df.info()
```

```python
# 🔍 Step 1: Find the problematic non-numeric values
float_columns = ['floors', 'waterfront', 'view', 'yr_built']

for col in float_columns:
    print(f"\n--- Checking {col} ---")
    bad_vals = df[~df[col].astype(str).str.replace('.', '', regex=False).str.isnumeric()][col].unique()
    print(bad_vals)
```

```python
for col in ['floors', 'waterfront', 'view', 'yr_built']:
    print(f"\n--- {col} ---")
    print(df[col].unique()[:20])  # show first 20 unique values
```

```python
#  Safely convert object/float columns to integer (Int64)
float_columns = ['floors', 'waterfront', 'view', 'yr_built']

for col in float_columns:
    # Step 1: Force numeric conversion
    df[col] = pd.to_numeric(df[col], errors='coerce')

    # Step 2: Round values before int conversion (avoid float errors)
    df[col] = df[col].round(0)

    # Step 3: Convert to Int64 (supports NaN)
    df[col] = df[col].astype('Int64')

print("Successfully converted to Int64:", float_columns)
print(df[float_columns].dtypes)
```

```python
df.info()
```

```python
# Convert all Int64 and int64 columns to plain int64 (standard lowercase)
for col in df.select_dtypes(include=['Int64', 'int64']).columns:
    df[col] = df[col].fillna(0).astype('int64')

print("All integer columns converted to standard int64:")
print(df.dtypes)
```
